# Correct a FITS spectrum with stellar exclusions

The template identifies stellar features to exclude from fitting; it is not subtracted from the observation. These velocities are illustrative Beta Pictoris settings. See [the template reference](../parameter_descriptions/TheoreticalSpectrum.md).


```python
from pathlib import Path

from pymolfit import TheoreticalSpectrum, correct

%matplotlib inline

GUIDE = Path("docs/user_guide") if Path("docs/user_guide").is_dir() else Path(".")
DATA = GUIDE / "data"
```

```python
template = TheoreticalSpectrum(
    path=DATA / "beta_pic_template.txt",
    radial_velocity_kms=20.0,
    vsini_kms=130.0,
    wavelength_unit="angstrom",
    wavelength_medium="air",
)
```

```python
result = correct(
    input_path=DATA / "harps_nad_crop_air.fits",
    theoretical_spectrum=template,
    output_path=GUIDE / "outputs/beta_pic_template.ecsv",
    show_plot=True,
)
```

Inspect exclusions in [the selector](interactive_selector.md) before relying on them. The bundled template is a padded optical crop, not a template for every wavelength.


```python
print(result.success, result.message)
```
