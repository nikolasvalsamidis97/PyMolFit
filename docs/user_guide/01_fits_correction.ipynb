# Correct a FITS spectrum

This Beta Pictoris Na D crop keeps the example quick. For full coverage, use `ADP.2017-04-07T01_04_41.632.fits` in the same data folder. Setup and data provenance are in [README.md](README.md).


```python
from pathlib import Path

from pymolfit import correct

%matplotlib inline

GUIDE = Path("docs/user_guide") if Path("docs/user_guide").is_dir() else Path(".")
DATA = GUIDE / "data"
```

```python
result = correct(
    input_path=DATA / "harps_nad_crop_air.fits",
    output_path=GUIDE / "outputs/beta_pic_fits.ecsv",
    show_plot=True,
)
```

`result` contains the corrected spectrum, transmission and diagnostics. The ECSV retains the full product. For manual fit/exclusion regions, follow [the selector guide](interactive_selector.md) and add `region_file=...` to the call.


```python
print(result.success, result.message)
```
