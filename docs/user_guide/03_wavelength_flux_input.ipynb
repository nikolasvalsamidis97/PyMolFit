# Correct a wavelength-flux file

The two-column text file contains the same HARPS crop: wavelength in nm (barycentric standard air) and flux in ADU. Its observing information must be supplied separately. Replace the example-specific values below with your observing log.


```python
from pathlib import Path

import numpy as np

from pymolfit import Observation, Spectrum, correct

%matplotlib inline

GUIDE = Path("docs/user_guide") if Path("docs/user_guide").is_dir() else Path(".")
DATA = GUIDE / "data"
```

```python
wavelength, flux = np.loadtxt(DATA / "beta_pic_nad.txt", unpack=True)

observation = Observation(
    time="2017-04-06T00:05:02.828",
    latitude_deg=-29.2584,
    longitude_deg=-70.7345,
    altitude_m=2400.0,
    airmass=1.2335,
    resolving_power=115000.0,
    wavelength_frame="barycentric",
    frame_velocity_km_s=-7.52937637262916,
)
```

```python
spectrum = Spectrum(
    wavelength=wavelength,
    flux=flux,
    wavelength_unit="nm",
    wavelength_medium="air",
    observation=observation,
)
```

When available, also supply `uncertainty=...`, `mask=...` (True = usable), and `group_id=...` on `Spectrum`. The barycentric correction above is not the stellar radial velocity.


```python
result = correct(
    input_path=spectrum,
    output_path=GUIDE / "outputs/beta_pic_arrays.ecsv",
    show_plot=True,
)
```

```python
print(result.success, result.message)
```
