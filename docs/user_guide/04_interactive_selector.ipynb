# Select, edit and apply fitting regions

Run these cells in order. Pause at the editor to adjust the regions before saving. Use a kernel with `pymolfit[interactive]` installed. Fitting regions constrain the atmosphere; exclusions protect stellar features and bad pixels.

```python
%matplotlib widget
from pathlib import Path

from pymolfit import correct, select_telluric_regions

GUIDE = Path("docs/user_guide") if Path("docs/user_guide").is_dir() else Path(".")
fits = GUIDE / "data/harps_nad_crop_air.fits"
region_file = GUIDE / "regions/beta_pic.ecsv"
```

```python
selector = select_telluric_regions(
    fits,
    automatic_region_count=100,
    automatic_region_half_width_pixels=12,
    output_path=region_file,
    show_interactive_window=True,
)
```

## Edit before continuing

Zoom to a feature. Choose **Fit** or **Exclude**, enable **Draw regions**, then drag over wavelength intervals. Disable drawing before panning. **Delete**, **Undo** and **Clear** change the selection. Exclusions override fitting regions.

The requested 100 lines can produce fewer merged windows. If `region_file` already exists, the editor reopens it instead of generating new intervals. To start fresh, omit `output_path` in the cell above. The **Automatic** button adds windows rather than replacing existing ones.

```python
selector.save(region_file)
```

## Correct using the saved regions

The ECSV records fitting and exclusion intervals, units, wavelength medium and reference frame. Regions affect parameter estimation, not correction coverage.

```python
result = correct(
    input_path=fits,
    region_file=region_file,
    output_path=GUIDE / "outputs/beta_pic_regions.ecsv",
    show_plot=True,
)
```

For automatic stellar exclusions, pass `theoretical_spectrum=template` to the selector, using the object from [notebook 2](02_fits_with_stellar_template.ipynb). Add `enable_theoretical_controls=True` to edit template and masking settings.

For generation without a window, omit `show_interactive_window=True` and still call `selector.save(region_file)`. See the [selector reference](interactive_selector.md) for saved-file reuse and time-series details.
